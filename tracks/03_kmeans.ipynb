# 03 k-means++

```python
import sys
import time
import numpy as np
sys.path.append("..")          # cf_common.py is in the folder above
from sklearn.cluster import KMeans
from cf_common import load_data, evaluate_clustering, save_results

data = load_data()
K_VALUES = [10, 20, 40, 80]
```

## The clustering method

```python
def cluster(points, k):
    return KMeans(n_clusters=k, n_init=3, random_state=579).fit_predict(points)
```

## Evaluate every k

```python
summaries, per_user = [], []
for k in K_VALUES:
    start = time.perf_counter()
    labels = cluster(data.embedding, k)
    fit_seconds = time.perf_counter() - start

    summary, users = evaluate_clustering(data, "kmeans", k, labels, fit_seconds)
    summaries.append(summary)
    per_user.append(users)
```

## Results

```python
results = save_results("kmeans", summaries, per_user)
results[["k", "clusters", "largest cluster %", "P@5", "P@10", "R@10", "P@20", "R@20",
         "clustering s", "CF s (all users, estimated)"]].round(4)
```
